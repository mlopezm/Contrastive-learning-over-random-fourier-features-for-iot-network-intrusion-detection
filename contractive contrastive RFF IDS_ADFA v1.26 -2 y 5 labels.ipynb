```python
#https://gist.github.com/mmmikael/0a3d4fae965bdbec1f9d  y https://github.com/keras-team/keras/blob/master/examples/mnist_siamese.py
def euclidean_distance(vects):
    x, y = vects
    sum_square = K.sum(K.square(x - y), axis=1, keepdims=True)
    return K.sqrt(K.maximum(sum_square, K.epsilon()))


def eucl_dist_output_shape(shapes):
    shape1, shape2 = shapes
    return (shape1[0], 1)

def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)

def dummy_loss(y_true, d): # d viene de la funcion exponencial y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0

def dummy_loss_max_margin(y_true, d): # d es la diferencia de distancias entre el valor x y el label verdadero y el mas cercano de los falsos [-Inf,+Inf]
  margin = 1
  return K.mean(K.maximum(margin + d, 0)) # y_true es siempre 0
```

### Modelo hinge loss multiclass

```python
# according to keras
def categorical_hinge(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = K.max((1. - y_true) * y_pred, axis=-1)
    return K.maximum(0., neg - pos + 1.)
```

```python
# according for multiclass SVM cost function to http://cs231n.github.io/linear-classify/
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
def categorical_hinge_multiclass_svm(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = (1. - y_true) * y_pred
    m = K.int_shape(y_pred)[1]
    pos_rep = K.repeat_elements(K.reshape(pos,(-1,1)),  m, axis=1)
    dif= neg-pos_rep
    dif_hinge=K.maximum(0., dif + 1.)
    dif_hinge_corrected=dif_hinge*(1. - y_true)
    dif_hinge_corrected_sum=K.sum(dif_hinge_corrected, axis=-1)
    return dif_hinge_corrected_sum
```

```python
# according for multiclass SVM cost function to http://cs231n.github.io/linear-classify/
# for the perceptron cost function https://www.cc.gatech.edu/~bboots3/CS4641-Fall2016/Lectures/Lecture5.pdf
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
def categorical_hinge_multiclass_perceptron(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = (1. - y_true) * y_pred
    m = K.int_shape(y_pred)[1]
    pos_rep = K.repeat_elements(K.reshape(pos,(-1,1)),  m, axis=1)
    dif= neg-pos_rep
    dif_hinge=K.maximum(0., dif )
    dif_hinge_corrected=dif_hinge*(1. - y_true)
    dif_hinge_corrected_sum=K.sum(dif_hinge_corrected, axis=-1)
    return dif_hinge_corrected_sum
```

# Preparar datos

```python
def MoverValDerecha(y):
    y_out = np.zeros_like((y))
    y_out[range(len(y_out)),(np.argmax(y,axis=1)+1)%rounds] = 1
    return(y_out)
```

```python
def ExtendX (m_ext,m):
    for i in range(rounds-1):
        m_ext = np.concatenate((m_ext,m), axis=0)
    return(m_ext)
```

```python
def ExtendY (m_ext,m):
    for i in range(rounds-1):
        m= MoverValDerecha(m)
        m_ext = np.concatenate((m_ext,m), axis=0)
    return(m_ext)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.1

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.00001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.000001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2429)
tf.random.set_seed(32773)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 14284
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.00001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(24215889)
tf.random.set_seed(3221973)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 123451
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 500
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

alpha_val = 0.0001

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 600
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 500
gamma = 3
seed = 1234
alpha_val = 0.00001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 100
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(h_cos)  # N_batch x N_hidden, por ser cosine

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 50
gamma = 3
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val,L2reg):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

### Preparar modelo

```python
np.random.seed(24215889)
tf.random.set_seed(3221973)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)

    return (logloss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val,L2reg):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.00001
L2reg = 0.1

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val,L2reg)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
#model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
model.compile(loss='binary_crossentropy', optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)

    return (logloss)

```

```python

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val,L2reg):
    
    tf.random.set_seed(seed)
    #init random weights
    ini_w_x = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_x = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    ini_w_l = tf.keras.initializers.RandomNormal(mean=0., stddev=np.sqrt(2 * gamma) )
    ini_b_l = tf.keras.initializers.RandomUniform(minval=0., maxval=2*np.pi)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_x, 
                     bias_initializer=ini_b_x,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "linear", trainable = True, 
                     kernel_initializer=ini_w_l, 
                     bias_initializer=ini_b_l,
                    kernel_regularizer=tf.keras.regularizers.l2(L2reg),
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
gamma = 3
seed = 1234
alpha_val = 0.00001
L2reg = 0.1

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, gamma, seed, alpha_val,L2reg)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
#model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
model.compile(loss='binary_crossentropy', optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2429)
tf.random.set_seed(32773)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = tf.math.sign(h_cos)  # N_batch x N_hidden, por ser relu

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)
```

```python
vocab_size = in_dim_label  # no usado de momento
embed_dim = 10

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val):
    
    tf.random.set_seed(seed)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = tf.math.sign(h_cos)  # N_batch x N_hidden, por ser relu

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)
```

```python
vocab_size = in_dim_label  # no usado de momento
embed_dim = 10

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val):
    
    tf.random.set_seed(seed)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

### Preparar modelo

```python
np.random.seed(2453115889)
tf.random.set_seed(322198573)
```

```python
def cosine_act(x):
    return tf.math.cos(x)
```

```python
def cosine_act_contracted_2(x):
    return (tf.math.cos(x) * np.sqrt(2.) / np.sqrt(rff_dim))
```

```python
def cosine_act_contracted_1(x):
    return (tf.math.cos(x) * 1. / np.sqrt(rff_dim))
```

```python
def sine_act_contracted_1(x):
    return (tf.math.sin(x) * 1. / np.sqrt(rff_dim))
```

```python
def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)
```

```python
def dummy_loss(y_true, d): # d es el valor que queremos minimizar y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0
```

```python
tf.compat.v1.disable_eager_execution()
def contractive_multi_loss(y_true, y_pred):

    logloss = tf.keras.metrics.binary_crossentropy(y_true, y_pred)
    
    #W = model.get_layer('contracted_layer').get_weights()[0]  # N x N_hidden
    W = K.variable(value=model.get_layer('layer_x').get_weights()[0])  # N x N_hidden
    W = K.transpose(W)  # N_hidden x N
    h_cos = model.get_layer('layer_x_out').output
    dh_cos = tf.math.sign(h_cos)  # N_batch x N_hidden, por ser relu

    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W**2, axis=1), axis=1))

    return (logloss + contractive_loss)
```

```python
vocab_size = in_dim_label  # no usado de momento
embed_dim = 10

def create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val):
    
    tf.random.set_seed(seed)
    
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))
    
    #RFF layers
    rff_x = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_x")
    rff_l = Dense(rff_dim, activation = "relu", trainable = True, 
                     name="layer_l")

    #embedding de x
    random_part_x = rff_x(in_x)
    latent_x = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_x_out")(random_part_x)
    
    
    #embedding de label
    random_part_l = rff_l(in_label)
    latent_l = tensorflow.keras.layers.Lambda(cosine_act_contracted_2, name="layer_l_out")(random_part_l)

    # similarity measure
    dot_product =  tensorflow.keras.layers.dot([latent_x, latent_l], 1, normalize=True)
    
    #  output layer
    #output = Dense(out_dim, activation='sigmoid')(dot_product)
    output = (dot_product+1.0)/2.0  # scale between 0-1
    
    #W = model.get_layer('layer_x').get_weights()[0]  # N x N_hidden
    W = rff_x.get_weights()[0]
    W2 = tf.convert_to_tensor(W, dtype=tf.float64)  # N x N_hidden
    W2 = K.transpose(W2)  # N_hidden x N
    #h_cos = model.get_layer('layer_x_out').output
    dh_cos = 1.0 - tf.math.square(latent_x)  # N_batch x N_hidden, por ser cosine
    # N_batch x N_hidden * N_hidden x 1 = N_batch x 1
    contractive_loss = alpha_val * (K.sum(dh_cos * K.sum(W2**2, axis=1), axis=1))

    
    # training model
    #m = Model(inputs=[in_x, in_label], outputs= [contractive_loss, output]) # devuelve salidas
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve salidas
    m_embed_x = Model(inputs=in_x, outputs= latent_x) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= latent_l) # devuelve embedding de label

    
    return(m, m_embed_x , m_embed_label, W2)
```

```python
#crear modelo
rff_dim = 400
seed = 1234
alpha_val = 0.0001

model, model_embed_x , model_embed_label, W2 = create_model_1(in_dim_x,in_dim_label,out_dim, rff_dim, seed, alpha_val)
#tf.compat.v1.enable_eager_execution()
#model.compile(loss=[dummy_loss, 'binary_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [1.0, 1.0] )
model.compile(loss=contractive_multi_loss, optimizer=Adam(0.001), metrics=['accuracy'])
```

## Comparacion de los valores medios de las features, training dataset

```python
x_embeddings_normal = model_embed_x.predict(XTrainM[YTrainANMOhe[:,1]==0]) # embeddings de las muestras normales
x_embeddings_anomaly = model_embed_x.predict(XTrainM[YTrainANMOhe[:,1]==1]) # embeddings de las muestras anomalas
```

```python
from scipy import stats
stat, pvalue = stats.ttest_1samp(np.abs(x_embeddings_normal_mean - x_embeddings_anomaly_mean), 0) 
print (stat, pvalue)
```
