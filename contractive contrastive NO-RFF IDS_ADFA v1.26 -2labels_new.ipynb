```python
#https://gist.github.com/mmmikael/0a3d4fae965bdbec1f9d  y https://github.com/keras-team/keras/blob/master/examples/mnist_siamese.py
def euclidean_distance(vects):
    x, y = vects
    sum_square = K.sum(K.square(x - y), axis=1, keepdims=True)
    return K.sqrt(K.maximum(sum_square, K.epsilon()))


def eucl_dist_output_shape(shapes):
    shape1, shape2 = shapes
    return (shape1[0], 1)

def contrastive_loss(y, d):
    """ Contrastive loss from Hadsell-et-al.'06
        http://yann.lecun.com/exdb/publis/pdf/hadsell-chopra-lecun-06.pdf
    """
    margin = 1
    square_pred = K.square(d)
    margin_square = K.square(K.maximum(margin - d, 0))
    return K.mean(y * square_pred + (1 - y) * margin_square)

def dummy_loss(y_true, d): # d viene de la funcion exponencial y su valor es [0,+inf]
  return K.mean(d - y_true) # y_true es siempre 0

def dummy_loss_max_margin(y_true, d): # d es la diferencia de distancias entre el valor x y el label verdadero y el mas cercano de los falsos [-Inf,+Inf]
  margin = 1
  return K.mean(K.maximum(margin + d, 0)) # y_true es siempre 0
```

```python
def accuracy(predictions, labels):
  return (np.sum(np.argmax(predictions, 1) == np.argmax(labels, 1))/ predictions.shape[0])

# given probability predictions return the one-hot encoded associated labels
def labelOHEncodedFromPrediction(predictions):
  return ((predictions == np.reshape(np.max(predictions, 1),(len(predictions),-1)))*1)

# given distance predictions return the one-hot encoded associated labels
def labelOHEncodedFromDistances(predictions):
  return ((predictions == np.reshape(np.min(predictions, 1),(len(predictions),-1)))*1)

#transform back labels (one-hot encoded) to strings for the case of labels with 4 values
def invTraADFAabels4labels(labels):
    return(lb.inverse_transform(labels))

#transform back labels (one-hot encoded) to strings for the case of labels with 2 values
def invTraADFAabels2labels(labels):
    return(np.array([ 'normal' if (x== [1,0]).all() else 'attack' for x in labels]))

#transform back labels (one-hot encoded) to strings for the case of labels with 2 values
def invTraADFAabels2labels_bis(labels):
    return(np.array([ 'normal' if x== [1,0] else 'attack' for x in labels]))

#transform labels to ohe for the case of labels with 2 values
def TraADFAabels2ohelabels(labels):
    return(np.array([[1,0] if x== 'normal' else [0,1] for x in labels]))

#transform labels to integers for the case of labels with 2 values
def TraADFAabels2Integerlabels(labels):
    return(np.array([0 if x== 'normal' else 1 for x in labels])) 
```

### Modelo hinge loss multiclass

```python
# according to keras
def categorical_hinge(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = K.max((1. - y_true) * y_pred, axis=-1)
    return K.maximum(0., neg - pos + 1.)
```

```python
# according for multiclass SVM cost function to http://cs231n.github.io/linear-classify/
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
def categorical_hinge_multiclass_svm(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = (1. - y_true) * y_pred
    m = K.int_shape(y_pred)[1]
    pos_rep = K.repeat_elements(K.reshape(pos,(-1,1)),  m, axis=1)
    dif= neg-pos_rep
    dif_hinge=K.maximum(0., dif + 1.)
    dif_hinge_corrected=dif_hinge*(1. - y_true)
    dif_hinge_corrected_sum=K.sum(dif_hinge_corrected, axis=-1)
    return dif_hinge_corrected_sum
```

```python
# according for multiclass SVM cost function to http://cs231n.github.io/linear-classify/
# for the perceptron cost function https://www.cc.gatech.edu/~bboots3/CS4641-Fall2016/Lectures/Lecture5.pdf
# y_true one-hot-encoded
# y_pred real values, the bigger value is the predicted answer
def categorical_hinge_multiclass_perceptron(y_true, y_pred):
    pos = K.sum(y_true * y_pred, axis=-1)
    neg = (1. - y_true) * y_pred
    m = K.int_shape(y_pred)[1]
    pos_rep = K.repeat_elements(K.reshape(pos,(-1,1)),  m, axis=1)
    dif= neg-pos_rep
    dif_hinge=K.maximum(0., dif )
    dif_hinge_corrected=dif_hinge*(1. - y_true)
    dif_hinge_corrected_sum=K.sum(dif_hinge_corrected, axis=-1)
    return dif_hinge_corrected_sum
```

### Modelo RBF

```python
#https://github.com/PetraVidnerova/rbf_keras
class InitCentersKMeans(Initializer):
    def __init__(self, X, max_iter=100):
        self.X = X
        self.max_iter = max_iter

    def __call__(self, shape, dtype=None):
        assert shape[1] == self.X.shape[1]

        n_centers = shape[0]
        km = KMeans(n_clusters=n_centers, max_iter=self.max_iter, verbose=0)
        km.fit(self.X)
        return km.cluster_centers_

class InitCentersRandom(Initializer):
    def __init__(self, X):
        self.X = X

    def __call__(self, shape, dtype=None):
        assert shape[1] == self.X.shape[1]
        idx = np.random.randint(self.X.shape[0], size=shape[0])
        return self.X[idx, :]


class RBFLayer(Layer):
    def __init__(self, output_dim, initializer=None, init_betas=1.0, **kwargs):
        self.output_dim = output_dim
        self.init_betas = init_betas
        if not initializer:
            self.initializer = RandomUniform(0.0, 1.0)
        else:
            self.initializer = initializer
        super(RBFLayer, self).__init__(**kwargs)

    def build(self, input_shape):

        self.centers = self.add_weight(name='centers',
                                       shape=(self.output_dim, input_shape[1]),
                                       initializer=self.initializer,
                                       trainable=True)
        self.betas = self.add_weight(name='betas',
                                     shape=(self.output_dim,),
                                     initializer=Constant(
                                         value=self.init_betas),
                                     trainable=True)

        super(RBFLayer, self).build(input_shape)

    def call(self, x):

        C = K.expand_dims(self.centers)
        H = K.transpose(C-K.transpose(x))
        return K.exp(-self.betas * K.sum(H**2, axis=1))

    def compute_output_shape(self, input_shape):
        return (input_shape[0], self.output_dim)
    
    def get_config(self):
        config = super().get_config().copy()
        config.update({
            'output_dim': self.output_dim,
            'init_betas': self.init_betas,
            'initializer': self.initializer,
        })
        return config
```

# Preparar datos

```python
def MoverValDerecha(y):
    y_out = np.zeros_like((y))
    y_out[range(len(y_out)),(np.argmax(y,axis=1)+1)%rounds] = 1
    return(y_out)
```

```python
def ExtendX (m_ext,m):
    for i in range(rounds-1):
        m_ext = np.concatenate((m_ext,m), axis=0)
    return(m_ext)
```

```python
def ExtendY (m_ext,m):
    for i in range(rounds-1):
        m= MoverValDerecha(m)
        m_ext = np.concatenate((m_ext,m), axis=0)
    return(m_ext)
```

### Preparar modelo

```python
np.random.seed(2455889)
tf.random.set_seed(32258573)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    dot_product =  dot([x_embed, label_embed], 1, normalize=True)
    
    # sigmoid output layer
    output = Dense(out_dim, activation='sigmoid')(dot_product)
    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss='binary_crossentropy', optimizer='rmsprop', metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
#crear modelo
model, model_embed_x , model_embed_label = create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    dot_product =  dot([x_embed, label_embed], 1, normalize=True)
    
    # sigmoid output layer
    output = Dense(out_dim, activation='sigmoid')(dot_product)
    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss='binary_crossentropy', optimizer='rmsprop', metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
#crear modelo
model, model_embed_x , model_embed_label = create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    u = Dense(embed_dim,activation='linear')(u)
    x_embed = Lambda(lambda x: tf.math.l2_normalize(x, axis=1),name="lambda_1")(u) # L2 normalize embeddings

    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    v = Dense(embed_dim,activation='linear')(v)
    label_embed = Lambda(lambda x: tf.math.l2_normalize(x, axis=1),name="lambda_2")(v) # L2 normalize embeddings

    # similarity measure
    dot_product =  dot([x_embed, label_embed], 1, normalize=False)
    
    # sigmoid output layer
    output = Dense(out_dim, activation='sigmoid')(dot_product)
    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss='binary_crossentropy', optimizer='rmsprop', metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
#crear modelo
model, model_embed_x , model_embed_label = create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim) 
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    distance = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed])

    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= distance) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss=contrastive_loss, optimizer=Adam(0.001), metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label_p = Input(shape=(in_dim_label,))
    in_label_n = Input(shape=(in_dim_label,))

    dd1 = Dense (50, activation='relu')
    dd2 = Dense (50, activation='relu')
    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label p
    v1 = dd1(in_label_p)
    v1 = dd2(v1)
    label_embed_p = Dense(embed_dim,activation='linear')(v1)
    #embedding de label n
    v2 = dd1(in_label_n)
    v2 = dd2(v2)
    label_embed_n = Dense(embed_dim,activation='linear')(v2)

    # similarity measure
    distance_p = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed_p])
    distance_n = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed_n])

    distance = distance_p - distance_n
    
    m_dist = tf.concat([distance_p, distance_n], axis=1)
    
    # training model
    m = Model(inputs=[in_x, in_label_p, in_label_n], outputs= distance) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label_p, outputs= label_embed_p) # devuelve embedding de label
    m_embed_distances = Model(inputs=[in_x, in_label_p, in_label_n], outputs= m_dist)
    m_embed_distance_p = Model(inputs=[in_x, in_label_p], outputs= distance_p)
    
    m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])
    return(m, m_embed_x , m_embed_label, m_embed_distances, m_embed_distance_p)
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 8

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label_p = Input(shape=(in_dim_label,))
    in_label_n = Input(shape=(in_dim_label,))

    dd1 = Dense (50, activation='relu')
    dd2 = Dense (50, activation='relu')
    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label p
    v1 = dd1(in_label_p)
    v1 = dd2(v1)
    label_embed_p = Dense(embed_dim,activation='linear')(v1)
    #embedding de label n
    v2 = dd1(in_label_n)
    v2 = dd2(v2)
    label_embed_n = Dense(embed_dim,activation='linear')(v2)

    # similarity measure
    distance_p = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed_p])
    distance_n = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed_n])

    distance = distance_p - distance_n
    
    m_dist = tf.concat([distance_p, distance_n], axis=1)
    
    # training model
    m = Model(inputs=[in_x, in_label_p, in_label_n], outputs= distance) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label_p, outputs= label_embed_p) # devuelve embedding de label
    m_embed_distances = Model(inputs=[in_x, in_label_p, in_label_n], outputs= m_dist)
    m_embed_distance_p = Model(inputs=[in_x, in_label_p], outputs= distance_p)
    
    m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])
    return(m, m_embed_x , m_embed_label, m_embed_distances, m_embed_distance_p)
```

```python
#crear modelo
model, model_embed_x , model_embed_label, model_mat_dist, model_dist_p = create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim) 
```

```python
vocab_size = in_dim_label  # no usado de momento
embed_dim = 4

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (100, activation='relu')(u)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    v = Dense (10, activation='relu')(v)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    dot_product =  dot([x_embed, label_embed], 1, normalize=True)
    
    # sigmoid output layer
    output = Dense(out_dim, activation='sigmoid')(dot_product)
    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss='binary_crossentropy', optimizer='rmsprop', metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
vocab_size = in_dim_label  # no usado de momento
embed_dim = 8

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (100, activation='relu')(u)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    v = Dense (10, activation='relu')(v)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    dot_product =  dot([x_embed, label_embed], 1, normalize=True)
    
    # sigmoid output layer
    output = Dense(out_dim, activation='sigmoid')(dot_product)
    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= output) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss='binary_crossentropy', optimizer='rmsprop', metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 10

def create_model_1(in_dim_x,in_dim_label,out_dim,embed_dim):
    # input variables
    in_x = Input(shape=(in_dim_x,))
    in_label = Input(shape=(in_dim_label,))

    
    #embedding1 = Embedding(vocab_size, embed_dim, name='embedding1')

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)
    
    #embedding de label
    v = Dense (10, activation='relu')(in_label)
    label_embed = Dense(embed_dim,activation='linear')(v)

    # similarity measure
    distance = Lambda(euclidean_distance,
                  output_shape=eucl_dist_output_shape)([x_embed, label_embed])

    
    # training model
    m = Model(inputs=[in_x, in_label], outputs= distance) # devuelve probabilidad
    m_embed_x = Model(inputs=in_x, outputs= x_embed) # devuelve embedding de x
    m_embed_label = Model(inputs= in_label, outputs= label_embed) # devuelve embedding de label

    
    m.compile(loss=contrastive_loss, optimizer=Adam(0.001), metrics=['accuracy'])
    return(m, m_embed_x , m_embed_label)
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(tf.exp(dist_to_true - min_dist_to_false),(-1,1))
    
    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss='mse', optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(tf.exp(dist_to_true - min_dist_to_false),(-1,1))
    
    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss='mse', optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(tf.exp(dist_to_true - min_dist_to_false),(-1,1))
    
    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=dummy_loss, optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias) # 

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_xembed_mas_distancias,a,b,c,d,e,f)
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out1 = tf.reshape( - min_dist_to_false,(-1,1))
    out2 = tf.reshape( dist_to_true,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= [out1,out2]) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias) # 

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    #m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])
    m.compile(loss=[dummy_loss_max_margin, dummy_loss], optimizer=Adam(0.001), metrics=['mse','mse'] , loss_weights = [0.5,0.5] )

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_xembed_mas_distancias,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_xembed_mas_distancias,a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 8

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out1 = tf.reshape( - min_dist_to_false,(-1,1))
    out2 = tf.reshape( dist_to_true,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= [out1,out2]) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias) # 

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    #m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])
    m.compile(loss=[dummy_loss_max_margin, dummy_loss], optimizer=Adam(0.001), metrics=['mse','mse'] , loss_weights = [0.5,0.5] )

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_xembed_mas_distancias,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_xembed_mas_distancias,a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 8

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= out) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias) # 

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=dummy_loss_max_margin, optimizer=Adam(0.001), metrics=['mse'])

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_xembed_mas_distancias,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_xembed_mas_distancias,a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # output de clasificacion
    n = Dense (20, activation='relu')(x_embed)
    n = Dense (20, activation='relu')(n)
    out_classification = Dense(in_dim_label,activation='softmax')(n)


    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= [out, out_classification]) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_out_classification = Model(inputs=in_x, outputs= out_classification)  # salida de solo la clasificacion

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=[dummy_loss_max_margin, 'categorical_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [0.5,0.5] )

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_out_classification,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_out_classification, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

## Model C-plet loss, 2 labels,  con dummy loss/max margin y loss de clasificacion adicional, usando las distancias como predictores adicionales, contrastive euclidean distance, dim 2; CEDist-NMS

```python
index_label_train = Ytrain_label
index_label_test = Ytest_label

list_labels = ['normal', 'attack']

print(TraADFAabels2ohelabels(['normal', 'attack']))
print(TraADFAabels2Integerlabels(list_labels))
print (index_label_train)

x_train, x_test, label_train, label_test, y_train, y_test = XTrainM, XTestM, \
                                                            YTrainANMOhe, YTestANMOhe,\
                                                            np.zeros(len(XTrainM)), np.zeros(len(XTestM))
print (x_train.shape, x_test.shape, label_train.shape, label_test.shape, y_train.shape, y_test.shape)
```

### Preparar modelo

```python
#np.random.seed(2453115889)
#tf.random.set_seed(322198573)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 2

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))
    # output de clasificacion
    n = Dense (50, activation='relu')(xembed_mas_distancias)
    n = Dense (50, activation='relu')(n)
    out_classification = Dense(in_dim_label,activation='softmax')(n)

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= [out, out_classification]) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_out_classification = Model(inputs=in_x, outputs= out_classification)  # salida de solo la clasificacion
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias)

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=[dummy_loss_max_margin, 'categorical_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [0.5,0.5] )

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_out_classification,m_xembed_mas_distancias,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_out_classification, model_xembed_mas_distancias, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
vocab_size = in_dim_label  # para los valores hasheados
embed_dim = 8

list_labels = ['normal', 'attack']

def create_all_labels(in_x):
  sal = tf.cast(tf.reshape(tf.tile(TraADFAabels2ohelabels(list_labels),
                                  [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
                                   dtype=tf.float64)
  return sal


def create_model_1(in_dim_x,in_dim_label,embed_dim):
    
    #input X
    in_x = Input(shape=(in_dim_x,))  
    
    # vector con enteros entre [0,in_dim_label-1] y de longitud igual a X.shape[0]. Indica cual es el indice del label correcto
    in_index_label = Input(shape=(1,),dtype=tf.int32) 
    
    #input labels, todos los labels, esta es una entrada construida aqui, no viene de fuera
    #in_labels = tf.cast(tf.reshape(tf.tile(lb5.transform(['DOS', 'NORMAL', 'PROBE', 'R2L', 'U2R']),
    #                              [tf.shape(in_x)[0],1]),[-1,in_dim_label,in_dim_label]), 
    #                               dtype=tf.float64)
    in_labels = Lambda(lambda x : create_all_labels(x))(in_x) # este rollo de crear un Lambda es para permitir salvar los pesos
    
    
    #listas de trabajo
    lst_labels=[]
    lst_dist=[]

    #embedding labels
    for i in range(in_dim_label):
        lst_labels.append(Dense(10, activation='relu')(in_labels[:,:,i]))
        lst_labels[i] = Dense(embed_dim, activation='linear')(lst_labels[i])

    #embedding de x
    u = Dense (100, activation='relu')(in_x)
    u = Dense (50, activation='relu')(u)
    x_embed = Dense(embed_dim,activation='linear')(u)

    # similarity measure
    for i in range(in_dim_label):
        lst_dist.append(Lambda(euclidean_distance,output_shape=eucl_dist_output_shape)([x_embed, lst_labels[i]]))

    mat_dist = tf.squeeze(tf.stack(lst_dist, axis=1))
    one_hot_mask = tf.reshape(tf.one_hot(in_index_label, in_dim_label, on_value = True, off_value = False, 
                                         axis = 1, dtype = tf.bool),(-1,in_dim_label))
    one_hot_mask_neg = tf.logical_not(one_hot_mask)
    dist_to_true = tf.boolean_mask(mat_dist, one_hot_mask)
    dist_to_false = tf.reshape(tf.boolean_mask(mat_dist, one_hot_mask_neg),(-1,in_dim_label-1))
    min_dist_to_false = tf.reduce_min(dist_to_false, axis = 1)
    out = tf.reshape(dist_to_true - min_dist_to_false,(-1,1))
    
    xembed_mas_distancias = tf.reshape(tf.concat([x_embed,mat_dist], axis = 1),(-1,embed_dim+in_dim_label))
    # output de clasificacion
    n = Dense (50, activation='relu')(xembed_mas_distancias)
    n = Dense (50, activation='relu')(n)
    out_classification = Dense(in_dim_label,activation='softmax')(n)

    # salida del modelo
    m = Model(inputs=[in_x,in_index_label],outputs= [out, out_classification]) # loss
    m_embed_x = Model(inputs=in_x, outputs= x_embed)  # embedding de x
    m_embed_labels = Model(inputs=in_x, outputs= lst_labels) #embedding de labels
    m_mat_dist = Model(inputs=in_x, outputs= mat_dist) # matriz de distancias entre x y labels
    m_lst_dist = Model(inputs=in_x, outputs= lst_dist) # # lista de distancias entre x y labels
    m_out_classification = Model(inputs=in_x, outputs= out_classification)  # salida de solo la clasificacion
    m_xembed_mas_distancias = Model(inputs=in_x, outputs= xembed_mas_distancias)

    a = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask)
    b = Model(inputs=[in_x,in_index_label],outputs= one_hot_mask_neg)
    c = Model(inputs=[in_x,in_index_label],outputs= mat_dist)
    d = Model(inputs=[in_x,in_index_label],outputs= dist_to_true)
    e = Model(inputs=[in_x,in_index_label],outputs= dist_to_false)
    f = Model(inputs=[in_x,in_index_label],outputs= min_dist_to_false)

    m.compile(loss=[dummy_loss_max_margin, 'categorical_crossentropy'], optimizer=Adam(0.001), metrics=['mse','accuracy'] , loss_weights = [0.5,0.5] )

    #return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist)
    return(m, m_embed_x , m_embed_labels, m_mat_dist, m_lst_dist,m_out_classification,m_xembed_mas_distancias,a,b,c,d,e,f)
```

```python
#crear modelo
model, model_embed_x , model_embed_labels, model_mat_dist, model_lst_dist, model_out_classification, model_xembed_mas_distancias, a,b,c,d,e,f = create_model_1(in_dim_x, in_dim_label, embed_dim) 
```
